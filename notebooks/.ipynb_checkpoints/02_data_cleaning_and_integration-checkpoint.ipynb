```python
import pandas as pd

# Load required datasets
customers = pd.read_csv("../olist/olist_customers_dataset.csv")
orders = pd.read_csv("../olist/olist_orders_dataset.csv")
order_items = pd.read_csv("../olist/olist_order_items_dataset.csv")
reviews = pd.read_csv("../olist/olist_order_reviews_dataset.csv")
products = pd.read_csv("../olist/olist_products_dataset.csv")

# Convert timestamp columns to datetime
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"]
)

print("Data loaded and timestamps converted.")
```

```python
# Merge customers with orders
cust_orders = customers.merge(
    orders,
    on="customer_id",
    how="inner"
)

print("Customer + Orders shape:", cust_orders.shape)
cust_orders.head()
```

```python
# Merge with order items (adds product & price)
cust_orders_items = cust_orders.merge(
    order_items,
    on="order_id",
    how="inner"
)

print("After adding order items:", cust_orders_items.shape)
cust_orders_items.head()
```

```python
# Merge reviews
cust_full = cust_orders_items.merge(
    reviews,
    on="order_id",
    how="left"
)

print("After adding reviews:", cust_full.shape)
cust_full.head()
```

```python
# Merge product categories
cust_full = cust_full.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

print("Final merged shape:", cust_full.shape)
cust_full.head()
```

```python
master_df = cust_full[[
    "customer_unique_id",
    "order_id",
    "order_purchase_timestamp",
    "order_status",
    "product_id",
    "price",
    "review_score",
    "review_comment_message",
    "product_category_name"
]]

print(master_df.shape)
master_df.head()
```

```python
# Save master customer table
master_df.to_csv(
    "../outputs/master_customer_table.csv",
    index=False
)

print("Master customer table created and saved.")
```
