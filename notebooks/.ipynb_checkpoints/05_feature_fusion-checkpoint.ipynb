```python
import pandas as pd

rfm_scaled = pd.read_csv("../outputs/rfm_scaled.csv")
nlp_features = pd.read_csv("../outputs/nlp_features.csv")

print("RFM shape:", rfm_scaled.shape)
print("NLP shape:", nlp_features.shape)

rfm_scaled.head()
```

```python
# Merge on customer_unique_id
X_full = rfm_scaled.merge(
    nlp_features,
    on="customer_unique_id",
    how="inner"
)

print("Fused feature shape:", X_full.shape)
X_full.head()
```

```python
X_full.to_csv(
    "../outputs/fused_features.csv",
    index=False
)

print("Feature fusion completed.")
```
