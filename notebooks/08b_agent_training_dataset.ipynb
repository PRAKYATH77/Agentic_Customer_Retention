```python
import pandas as pd

rfm = pd.read_csv("../outputs/rfm_raw.csv")
nlp = pd.read_csv("../outputs/nlp_features.csv")
personas = pd.read_csv("../outputs/customer_personas.csv")
agent_actions = pd.read_csv("../outputs/agent_actions.csv")
```

```python
data = (
    agent_actions
    .merge(rfm, on="customer_unique_id", how="left")
    .merge(nlp, on="customer_unique_id", how="left")
)

data.head()
```

```python
data = data.dropna()
data.shape
```

```python
data.to_csv(
    "../outputs/agent_training_dataset.csv",
    index=False
)

print("agent_training_dataset.csv created.")
```
